# 🔗 DQL - Data Query Language & JOINs

O comando `SELECT` é utilizado para consultar e filtrar informações no banco de dados.

#### 1. Consulta e Ordem de Escrita (Sintaxe)
O livro *SQL - Guia Prático* (Alice Zhao) faz uma brincadeira (mnemonica) bem legal para ajudar a decorar essa ordem:
> *"Start Fridays With Grandma's Homemade Oatmeal"* (**S-F-W-G-H-O**)

1. `SELECT` — Colunas a serem exibidas
2. `FROM` — Tabela principal
3. `WHERE` — Filtro de linhas brutas
4. `GROUP BY` — Agrupamento
5. `HAVING` — Filtro de dados agrupados
6. `ORDER BY` — Ordenação do resultado

Ordem de Execução Lógica no Banco:
`FROM` -> `WHERE` -> `GROUP BY` -> `HAVING` -> `SELECT` -> `ORDER BY`

#### 2. Junções (`JOINs`)

JOINs são usados para combinar colunas de duas ou mais tabelas com base em um relacionamento entre elas.

* **`INNER JOIN`:** Retorna apenas correspondências exatas em ambas as tabelas.
* **`LEFT JOIN`:** Retorna todos os registros da tabela da esquerda e os correspondentes da direita.
* **`RIGHT JOIN`:** Retorna todos os registros da tabela da direita e os correspondentes da esquerda.
* **`FULL OUTER JOIN`:** Retorna todos os registros quando houver correspondência em qualquer uma das tabelas.

#### 3. Funções Agregadas, `GROUP BY` e `HAVING`

* `COUNT()`: Conta registros
* `SUM()`: Soma valores
* `AVG()`: Média aritmética
* `MAX()` / `MIN()`: Valor máximo / mínimo

> **Diferença:** `WHERE` filtra linhas **antes** de agrupar; `HAVING` filtra os grupos **depois** de calculadas as agregações.

#### Estrutura Geral de Consulta SQL (`SELECT`)

```python
-- Sintaxe: Consulta completa com agrupamento e filtros
SELECT 
    <tabela_a>.<coluna_agrupada>,
    FUNCAO_AGREGACAO(<tabela_b>.<coluna_valor>) AS <apelido_resultado>
FROM <tabela_a>
WHERE <tabela_a>.<condicao_linha>
GROUP BY <tabela_a>.<coluna_agrupada>
HAVING FUNCAO_AGREGACAO(<tabela_b>.<coluna_valor>) <condicao_grupo>
ORDER BY <apelido_resultado> ASC|DESC;
```

#### Sintaxe de Tipos de JOIN

```python
-- INNER JOIN: Apenas correspondências em A e B
SELECT A.<coluna1>, B.<coluna2>
FROM <tabela_a> A
INNER JOIN <tabela_b> B ON A.<chave_pk> = B.<chave_fk>;

-- LEFT JOIN: Todos de A, correspondentes em B
SELECT A.<coluna1>, B.<coluna2>
FROM <tabela_a> A
LEFT JOIN <tabela_b> B ON A.<chave_pk> = B.<chave_fk>;

-- RIGHT JOIN: Todos de B, correspondentes em A
SELECT A.<coluna1>, B.<coluna2>
FROM <tabela_a> A
RIGHT JOIN <tabela_b> B ON A.<chave_pk> = B.<chave_fk>;

-- FULL OUTER JOIN: Todos os registros de A e B
SELECT A.<coluna1>, B.<coluna2>
FROM <tabela_a> A
FULL OUTER JOIN <tabela_b> B ON A.<chave_pk> = B.<chave_fk>;
```
