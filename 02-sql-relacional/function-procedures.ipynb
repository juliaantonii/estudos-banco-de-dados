# 🔗 Stored Procedures e Functions (PL/pgSQL)

Em SGBDs relacionais como o **PostgreSQL**, podemos programar blocos de código diretamente no banco de dados para automatizar regras de negócio.

#### Diferença Principal:
* **`FUNCTION`:** Obrigatoriamente **retorna um valor** ou uma tabela. Pode ser chamada dentro de comandos `SELECT`.
* **`PROCEDURE`:** Executada com o comando `CALL`. Usada para processamento de rotinas e controle transacional (`COMMIT`/`ROLLBACK`).

#### Sintaxe de Funções (`FUNCTION`)

```python
-- Sintaxe: Criar ou substituir Function
CREATE OR REPLACE FUNCTION <nome_funcao>(<parametro1> <TIPO_DADO>)
RETURNS <TIPO_RETORNO>
LANGUAGE plpgsql
AS $$
DECLARE
    <variavel_local> <TIPO_DADO>;
BEGIN
    SELECT <coluna> 
    INTO <variavel_local> 
    FROM <nome_tabela> 
    WHERE <coluna_filtro> = <parametro1>;
    
    RETURN <variavel_local>;
END;
$$;

-- Execução da Function
SELECT <nome_funcao>(<valor_parametro>);
```

#### Sintaxe de Stored Procedures (`PROCEDURE`)

```python
-- Sintaxe: Criar ou substituir Procedure
CREATE OR REPLACE PROCEDURE <nome_procedure>(
    <parametro1> <TIPO_DADO>,
    <parametro2> <TIPO_DADO>
)
LANGUAGE plpgsql
AS $$
BEGIN
    INSERT INTO <nome_tabela> (<coluna1>, <coluna2>)
    VALUES (<parametro1>, <parametro2>);
    
    -- Lógica adicional ou confirmação de transação
END;
$$;

-- Execução da Procedure
CALL <nome_procedure>(<valor1>, <valor2>);
```
