# 🔗 DDL - Data Definition Language (Linguagem de Definição de Dados)

---

A **DDL** é o subconjunto da linguagem SQL responsável por **definir, alterar e excluir a estrutura** das tabelas e do banco de dados (o esquemático).

#### 1. Principais Comandos:
* `CREATE`: Cria novos objetos (bancos, tabelas, índices).
* `ALTER`: Modifica a estrutura de um objeto existente.
* `DROP`: Remove permanentemente um objeto do banco de dados.
* `TRUNCATE`: Esvazia uma tabela mantendo sua estrutura intacta.

#### 2. Restrições de Integridade (Constraints)
* **`PRIMARY KEY` (PK):** Identificador único da linha. Não permite valores nulos ou duplicados.
* **`FOREIGN KEY` (FK):** Garante a integridade referencial entre duas tabelas.
* **`NOT NULL`:** Impede que a coluna receba valores nulos.
* **`UNIQUE`:** Garante que todos os valores na coluna sejam distintos.
* **`CHECK`:** Valida uma condição lógica antes de inserir/alterar o dado.

```python
-- Sintaxe: Criar tabela com Primary Key, Foreign Key e Constraints
CREATE TABLE <nome_tabela_pai> (
    <coluna_pk> <TIPO_DADO> PRIMARY KEY,
    <coluna_campo> <TIPO_DADO> NOT NULL UNIQUE
);

CREATE TABLE <nome_tabela_filha> (
    <coluna_pk> <TIPO_DADO> PRIMARY KEY,
    <coluna_campo> <TIPO_DADO> NOT NULL,
    <coluna_fk> <TIPO_DADO> NOT NULL,
    <coluna_valor> <TIPO_DADO> CHECK (<coluna_valor> > 0),
    
    FOREIGN KEY (<coluna_fk>) 
    REFERENCES <nome_tabela_pai>(<coluna_pk>)
);
```

#### Alteração e Remoção de Estruturas (`ALTER`, `DROP` e `TRUNCATE`)

```python
-- Sintaxe: Adicionar nova coluna
ALTER TABLE <nome_tabela> 
ADD COLUMN <nome_coluna> <TIPO_DADO>;

-- Sintaxe: Alterar tipo de dado de uma coluna
ALTER TABLE <nome_tabela> 
ALTER COLUMN <nome_coluna> TYPE <NOVO_TIPO_DADO>;

-- Sintaxe: Remover coluna
ALTER TABLE <nome_tabela> 
DROP COLUMN <nome_coluna>;

-- Sintaxe: Esvaziar dados mantendo a tabela
TRUNCATE TABLE <nome_tabela>;

-- Sintaxe: Excluir a tabela permanentemente
DROP TABLE <nome_tabela>;
```
