# 🔗 DML - Data Manipulation Language (Linguagem de Manipulação de Dados)

A **DML** lida diretamente com os **dados/registros** dentro das tabelas.

#### 1. Principais Comandos:
* `INSERT`: Insere novas linhas.
* `UPDATE`: Modifica dados existentes.
* `DELETE`: Remove linhas específicas.

> **SEMPRE** utilize a cláusula `WHERE` em comandos `UPDATE` e `DELETE` para evitar alterar ou apagar a base de dados inteira por engano!

#### Inserção de Dados (`INSERT`)

```python
-- Sintaxe: Inserir um único registro
INSERT INTO <nome_tabela> (<coluna1>, <coluna2>, <coluna3>) 
VALUES (<valor1>, <valor2>, <valor3>);

-- Sintaxe: Inserir múltiplos registros
INSERT INTO <nome_tabela> (<coluna1>, <coluna2>) 
VALUES 
    (<valor1_a>, <valor2_a>),
    (<valor1_b>, <valor2_b>),
    (<valor1_c>, <valor2_c>);
```

#### Atualização e Exclusão (`UPDATE` e `DELETE`)

```python
-- Sintaxe: Atualizar um ou mais campos
UPDATE <nome_tabela>
SET <coluna1> = <novo_valor1>,
    <coluna2> = <novo_valor2>
WHERE <condicao_filtro>;

-- Sintaxe: Deletar registros específicos
DELETE FROM <nome_tabela>
WHERE <condicao_filtro>;
```
